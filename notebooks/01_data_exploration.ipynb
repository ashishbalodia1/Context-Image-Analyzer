# 🌍 City Guardian — Exploratory Data Analysis (EDA)
### Multi-Domain Environmental Image Intelligence

This notebook provides automated EDA and diagnostic visualizations for citizen-captured environmental photos:
1. **Class Distribution**: Identifies class frequencies, rare violations, and extreme imbalances.
2. **Bounding Box Geometry**: Inspects normalized widths, heights, and aspect ratio profiles.
3. **Scale Analysis**: Categorizes objects into Small (<0.05), Medium (0.05 - 0.25), and Large (>0.25).
4. **Spatial Density Heatmaps**: Detects camera center-bias and framing artifacts.

```python
import os
import sys
from pathlib import Path

# Add project root to sys.path
project_root = Path('..').resolve()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.preprocessing.eda_analyzer import DatasetEDAAnalyzer
from src.preprocessing.image_enhancer import get_domain_augmentation_hyperparameters
print('EDA modules imported successfully!')
```

## 1. Select Environmental Domain & Run Automated Analysis

```python
# Choose domain: 'air_pollution', 'water_quality', 'sewage_drainage', 'mining_dust', 'solid_waste'
DOMAIN = 'solid_waste'
CONFIG_PATH = project_root / 'models' / 'configs' / f'{DOMAIN}.yaml'
LABELS_DIR = project_root / 'data' / 'processed' / DOMAIN / 'labels' / 'train'

analyzer = DatasetEDAAnalyzer(str(CONFIG_PATH))
metrics = analyzer.analyze_dataset(LABELS_DIR)

print(f'Total Images: {metrics.get("total_images", 0)}')
print(f'Total Boxes:  {metrics.get("total_boxes", 0)}')
print('Class Distribution:', metrics.get('class_distribution', {}))
print('Bounding Box Scales:', metrics.get('box_scales', {}))
```

## 2. Render Statistical Diagnostic Plots

```python
plots_dir = project_root / 'reports' / 'eda'
plots = analyzer.generate_eda_plots(metrics, plots_dir, domain_name=DOMAIN)
print(f'Generated {len(plots)} diagnostic plots:')
for p in plots:
    print(f'  - {p}')
```

## 3. Inspect Domain-Specific Augmentation Hyperparameters
Environmental phenomena (e.g. smoke, sewage, garbage) require specialized augmentation bounds to prevent unphysical transformations (e.g., zero vertical flip for smoke).

```python
hyp = get_domain_augmentation_hyperparameters(DOMAIN)
print(f'YOLOv8 Augmentation Profile for [{DOMAIN.upper()}]:')
for k, v in hyp.items():
    print(f'  {k:15s}: {v}')
```
