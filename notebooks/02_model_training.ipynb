# 🚀 City Guardian — YOLOv8 Model Fine-Tuning
### Multi-Domain Environmental Intelligence Engine

This notebook trains domain-specialist YOLOv8 detectors for:
1. **AQI & Air Pollution** (industrial smoke, open burning, dust clouds)
2. **Water Quality** (discoloration, floating waste, oil slick)
3. **Sewage & Drainage** (overflowing drains, blocked drains, stagnant wastewater)
4. **Mining Dust & Industrial** (dust plumes, uncovered mineral transport, excavation)
5. **Solid Waste** (illegal dumping, overflowing dustbins, plastic mounds)

```python
import sys
from pathlib import Path
import torch

# Set project root
project_root = Path('..').resolve()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from scripts.train_model import train_environmental_model
print(f'PyTorch Version: {torch.__version__}')
print(f'CUDA Available:  {torch.cuda.is_available()}')
```

## 1. Launch Domain Model Fine-Tuning

```python
# Select target domain
DOMAIN = 'solid_waste'  # or 'air_pollution', 'water_quality', 'sewage_drainage', 'mining_dust'

train_summary = train_environmental_model(
    domain=DOMAIN,
    model_size='s',        # 'n', 's', 'm'
    epochs=50,
    batch_size=16,
    optimizer='AdamW',
    lr0=0.001,
    patience=15,
    cos_lr=True
)
print('Training run completed:', train_summary['run_directory'])
```
